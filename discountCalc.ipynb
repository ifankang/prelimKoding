## Discount Calculator

```python
price=float(input("Original price :"))
discount=float(input("discount (%):"))
discount_amount=price*discount/100
final_price=price-discount_amount
print(f"Original price : {price:.2f}")
print(f"Discount amount : {discount_amount:,.2f}")
print(f"Final price is : Rp{final_price:,.2f}")
```

```python
from ipywidgets import FloatText, Button, VBox, HTML
from IPython.display import display

# Input untuk harga barang
price = FloatText(description='Price')

# Input untuk diskon
discount = FloatText(description='Discount %')

# Tempat menampilkan hasil
result = HTML()

# Tombol hitung
button = Button(description='Calculate')

# Fungsi yang dijalankan saat tombol diklik
def calculate(b):

    # Menghitung jumlah diskon
    discount_amount = (
        price.value * discount.value / 100
    )

    # Menghitung harga akhir
    final_price = (
        price.value - discount_amount
    )

    # Menampilkan hasil
    result.value = f"""
      <h3>Result</h3>

      <b>Original Price:</b>
      Rp {price.value:,.2f}

      <br><br>

      <b>Discount Amount:</b>
      Rp {discount_amount:,.2f}

      <br><br>

      <b>Final Price:</b>
      Rp {final_price:,.2f}
    """

# Hubungkan tombol dengan fungsi calculate
button.on_click(calculate)

# Susun komponen secara vertikal
app = VBox([
    price,
    discount,
    button,
    result
])

# Tampilkan aplikasi
display(app)
```
